```python

import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Student": ["Rukmini", "Asha", "Rahul", "Priya", "Amit"],
    "Study_Hours": [2, 3, 4, 5, 6],
    "Marks": [45, 60, 70, 85, 95]
}

df = pd.DataFrame(data)

print(df)
```

```python

print(df.describe())
print(df.isnull().sum())
```

```python

plt.bar(df["Student"], df["Marks"])
plt.title("Student Marks Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python

# Mini Project Summary

1. Students who studied more hours generally scored higher marks.
2. The dataset contains no missing values.
3. The highest marks are 95.
4. The average marks are calculated using Pandas.
5. The bar chart clearly shows the performance of each student.
```
